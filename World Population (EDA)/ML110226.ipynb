```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("world_population_data.csv")
df.head()
```

```python
df.columns
```

```python
df.rename(columns= {"density (km²)": "density"},inplace=True)
df.rename(columns= {"area (km²)": "area"},inplace=True)
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
sns.histplot(df["2023 population"], bins = 10, kde = True)
plt.title("Distrubition of 2023 Population")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
df.groupby("continent")["2023 population"].mean().sort_values(ascending=False).head(5)
```

```python
top_10_countries = df.sort_values(by = "2023 population", ascending=False).head(10)
```

```python
sns.barplot(x = "2023 population", y = "country", data = top_10_countries)
plt.title("Top 10 Countries by population in 2023")
plt.xlabel("Population")
plt.ylabel("Countries")
plt.show()
```

```python
sns.scatterplot(x = "2023 population", y = "density", data = df)
plt.title("Population of 2023 Density scatter plot")
plt.xlabel("Population")
plt.ylabel("Denstiy (km2)")
plt.show()
```

```python
df.groupby("continent")["area"].sum().sort_values(ascending=False)
```

```python
sns.violinplot(x="continent", y="density", data=df, palette="rainbow")
plt.title("Population Density distribution accross Continents")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show
```

```python
sns.boxplot(x="continent", y="density", data=df, palette="viridis",showfliers=False)
plt.title("Population Density distribution accross Continents (Using Outliers)")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show
```
