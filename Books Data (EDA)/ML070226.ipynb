```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#improting tools
```

```python
df = pd.read_csv("Books_Data_Clean.csv")
#reading data
```

```python
# reading top data
df.head()
```

```python
df.describe()
```

```python
df = df[df["Publishing Year"] >= 1900]
```

```python
df.isna().sum()
#checking nan/null
```

```python
df.dropna(subset= "Book Name", inplace= True)
#dropping null book names
```

```python
df.duplicated().sum()
#checking duplicates
```

```python
df.nunique()
```

```python
df.columns
```

```python
#Histogram showing the frequency of books published each year (adding colour)
plt.hist(df["Publishing Year"], color="orange")
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Years")
plt.show()
```

```python
#Histogram sum of total books for each genre (applying multiple colours)
df["genre"].value_counts().plot(kind = "bar", color = ["skyblue", "lightblue", "dodgerblue", "steelblue"])
plt.xlabel("Genre")
plt.ylabel("Number of Books")
plt.title("Number of books in each genre")
plt.show()
```

```python
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending=False)
# head to check top, tail to check bottom, reset index for different view
```

```python
#Boxplot (rating count for each genre), applying colour palettes
sns.boxplot(x = "genre", y = "Book_ratings_count", data = df, palette="Set3")
plt.xlabel("Genre")
plt.ylabel("Books rating count")
plt.title("Box plot of book rating count for each genre")
plt.show()
```

```python
#Scatter plot (sale price vs untis sold),applying gradient colours
plt.scatter(df["sale price"], df["units sold"], c=df["units sold"], cmap="rainbow")
plt.xlabel("Sale Price")
plt.ylabel("Units Sold")
plt.title("Scatter Plot of sale price vs units sold")
plt.show()
```

```python
language_counts = df["language_code"].value_counts().sort_values(ascending = True)
```

```python
#Piechart (language distribution), adding multiple colours with cmap
plt.pie(language_counts, labels = language_counts.index, startangle=90, autopct="%1.1f%%", colors=plt.cm.Set1.colors)
plt.title("Languange Distribution of books")
plt.show()
```

```python
df.groupby("Publisher ")["publisher revenue"].sum().sort_values(ascending= False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values( ascending= False).max()
```

```python
df.groupby("language_code").size().sort_values(ascending=False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].median()
```

```python
#Scatterplot (average book rating and book ratings)
plt.scatter(df["Book_average_rating"], df["Book_ratings_count"], c=df["Book_average_rating"], cmap = "Oranges")
plt.xlabel("Average books rating")
plt.ylabel("Books rating counts")
plt.title("Scatter of plot of book average rating vs book rating count")
plt.show
```

```python
totat__gross_sales_by_author = df.groupby("Author")["gross sales"].sum()
```

```python
#Histogram (Author and Total gross sales )
totat__gross_sales_by_author.sort_values(ascending=False).head(15).plot(kind = "bar", color=plt.cm.tab10.colors)
plt.xlabel("Author")
plt.ylabel("Total Gross Sales")
plt.title("Total Gross Sales for each Author")
plt.show()
```

```python
#Boxplot (Units sold and Author Rating relationship)
sns.boxplot(x="Author_Rating", y = "units sold", data= df, color="crimson")
plt.xlabel("Author Rating")
plt.ylabel("Units Sold")
plt.title("Boxplot of Units Sold from Author Rating")
plt.show
```

```python
#Linechart (Units Sold over the years)
df.groupby("Publishing Year")["units sold"].sum().plot(kind = "line", marker = "o", color="lightgreen")
plt.xlabel("Publishing Year")
plt.ylabel("Total units Sold")
plt.title("Total Units Sold Over the years")
plt.show()
```
